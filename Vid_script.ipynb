`1. importing Libraries`

```python
import cv2
import numpy as np 
import mediapipe as mp 
import os
import time
```

`2. creating instance of mp`

```python
holistic = mp.solutions.holistic
drawing = mp.solutions.drawing_utils 
# as it seems the soluitons api is depriciated there is a new api called tasks but we still opt to use the older version as there are more resources online for it 
```

```python

def mediapipe_detection(image, model):
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB) 
    # COLOR CONVERSION BGR 2 RGB (cuz open CV can only work with BGR but mediapipe uses RGB)
    
    image.flags.writeable = False               
    # Here image is a numpy array to be supplied in the future and when we turn of the writable flag , the array becomes locked  and becomes read only. 
    # locking is for performance optimization 
    # when mediapipe uses the  .process() function , it uses C++ code which cannot trust the array on non mutation in mid processing thus makes a copy and works on that copy. But locking the wriatbility eliminates the risk of mutation entirely.
    
    results = model.process(image)                 
    # Make predictions bout the landmarks 
    print("pose:", len(results.pose_landmarks.landmark) if results.pose_landmarks else 0)
    print("right hand:", len(results.right_hand_landmarks.landmark) if results.right_hand_landmarks else 0)
    print("left hand:", len(results.left_hand_landmarks.landmark) if results.left_hand_landmarks else 0)

    
    image.flags.writeable = True                   
    # Image is now unlocked to be  writable again 
    
    image = cv2.cvtColor(image, cv2.COLOR_RGB2BGR) 
    
    # COLOR CONVERSION RGB 2 BGR as the image is handed back to open CV
    return image, results
  
def draw_landmarks(image, results):
    # drawing.draw_landmarks(
    #   image, results.face_landmarks, holistic.FACE_CONNECTIONS) # Draw face connections
    # we dont need the facial landmarks 
    
    #Pose
    drawing.draw_landmarks(
      image, results.pose_landmarks, holistic.POSE_CONNECTIONS) # Draw pose connections
    
    #left hand
    drawing.draw_landmarks(
      image, results.left_hand_landmarks, holistic.HAND_CONNECTIONS) # Draw left hand connections
    
    #Right hand
    drawing.draw_landmarks(
      image, results.right_hand_landmarks, holistic.HAND_CONNECTIONS) # Draw right hand connections
    
def draw_styled_landmarks(image, results):
    # # Draw face connections
    # drawing.draw_landmarks(
    #   image, results.face_landmarks, holistic.FACE_CONNECTIONS,
    #   drawing.DrawingSpec(color=(80,110,10), thickness=1, circle_radius=1), 
    #   drawing.DrawingSpec(color=(80,256,121), thickness=1, circle_radius=1)) 
    # we dont need the face landmarks 

    # Draw pose connections
    drawing.draw_landmarks(image, results.pose_landmarks, holistic.POSE_CONNECTIONS,
                             drawing.DrawingSpec(color=(80,22,10), thickness=2, circle_radius=4), 
                             drawing.DrawingSpec(color=(80,44,121), thickness=2, circle_radius=2)
                             ) 
    # Draw left hand connections
    drawing.draw_landmarks(image, results.left_hand_landmarks, holistic.HAND_CONNECTIONS, 
                             drawing.DrawingSpec(color=(121,22,76), thickness=2, circle_radius=4), 
                             drawing.DrawingSpec(color=(121,44,250), thickness=2, circle_radius=2)
                             ) 
    # Draw right hand connections  
    drawing.draw_landmarks(image, results.right_hand_landmarks, holistic.HAND_CONNECTIONS, 
                             drawing.DrawingSpec(color=(245,117,66), thickness=2, circle_radius=4), 
                             drawing.DrawingSpec(color=(245,66,230), thickness=2, circle_radius=2)
                             ) 
```

`3. Live Webcam Feed`


```python
live_feed = cv2.VideoCapture(0)

with holistic.Holistic(min_detection_confidence=0.5, min_tracking_confidence=0.5) as holistic_instance:
    while True:
        ret,image=live_feed.read() # feed the live feed to cv2 client  ret is a flag used for clean exit 
        
        if ret == False: # use the ret flag for a clean exit 
            print("Failed to grab frame")
            break

        image, results = mediapipe_detection(image, holistic_instance)
        print("landmarks:", results) 

        draw_styled_landmarks(image,results) 

        cv2.imshow("Web cam live",image) # feeds the next image in the queue of images captured by live feed . Feeds it to the GUI 
        
        k=cv2.waitKey(1) # This sets the delay for repainting of the gui window. wait 1ms to check for any key press and assign it to k if no press k = -1

        if k==ord('q'): # ord = retuns the numeircal val the asci code of character
            break #holds the screen untill user closes the window

    live_feed.release()
    cv2.destroyAllWindows() #closes the window after user decides to
```
